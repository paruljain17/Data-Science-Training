```python
words = ["plays","playing","played",
         "run","running","happen",
         "happening","happen","history"]
```

```python
## porterstemmer
from nltk.stem import PorterStemmer
ps = PorterStemmer()

for word in words:
    print(word,"-------->",ps.stem(word))
```

```python
## regex stemmer -- regex is a pattern matching language
from nltk.stem import RegexpStemmer
rs = RegexpStemmer("ing$|s$|ed$|able$|ibles$")
for word in words:
    print(word,"--------->",rs.stem(word))
```

```python
#snowballstemmer --- updated version of portstemmer 
from nltk.stem import SnowballStemmer
ss=SnowballStemmer("english")
for word in words:
    print(word,"----->",ss.stem(word))
```
