```python
#tokenization 
# Tokenization is the process of splitting text into smaller units called tokens, which can be:
# Words
# Sentences
# Subwords (in advanced model)
```

```python
corpus = "Color (or colour in Commonwealth English; see spelling differences) is the visual perception"
```

```python
from nltk.tokenize import sent_tokenize, word_tokenize
```

```python
doc = sent_tokenize(corpus)
for i in doc:
    print(i)
    print('\n')
    print(word_tokenize(i))
    print('\n')
```

```python
##word punct tokenization -- . , ! ? : 
##punctuation tokenization
from nltk.tokenize import wordpunct_tokenize
wordpunct_tokenize(corpus)
```

```python
#treebank tokenization----more accurate
from nltk.tokenize import TreebankWordTokenizer
tokenizer=TreebankWordTokenizer()
tokenizer.tokenize(corpus)
```
