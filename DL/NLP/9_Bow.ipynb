```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
docs = ["The sun is shining ", "the wheteher is sweet" , 
        "The whether is sweet, The wheteher is sweet , and one and"]
```

```python
vec = CountVectorizer()
x = vec.fit_transform(docs)
# to see voculabry
print(vec.vocabulary_)
print(vec.get_feature_names_out())

print(x.toarray())
```
